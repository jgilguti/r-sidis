```python
import pandas as pd

# Load the original file
df = pd.read_csv("runs_times.csv")

# Convert start and stop times to datetime
df["start_time"] = pd.to_datetime(df["start_time"])
df["stop_time"] = pd.to_datetime(df["stop_time"])

# Compute fraction of an hour elapsed
df["length_h"] = (df["stop_time"] - df["start_time"]).dt.total_seconds() / 3600

# Save to new CSV
df.to_csv("new_run_times.csv", index=False)

print("✅ File saved as 'runs_times_with_length.csv'")
```
