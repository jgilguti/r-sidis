```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import os

plt.style.use('default')

df = pd.read_csv('cher_cal_results.csv')

four_pass_neg = df[(df['ebeam']==8.5831) & (df['thpq']==2.0) & (df['polarity']<0)]
four_pass_pos = df[(df['ebeam']==8.5831) & (df['thpq']==2.0) & (df['polarity']>0)]

five_pass_neg = df[(df['ebeam']==10.6716) & (df['thpq']==2.0) & (df['polarity']<0)]
five_pass_pos = df[(df['ebeam']==10.6716) & (df['thpq']==2.0) & (df['polarity']>0)]

plt.figure(figsize=(12, 6))

plt.errorbar(four_pass_neg['z'], four_pass_neg['hmscer_mean'], yerr=four_pass_neg['hmscer_err'], fmt='o', label = r"4-pass data, $\pi^-$", color='k', elinewidth=1.5, capsize=0)
plt.errorbar(four_pass_pos['z'], four_pass_pos['hmscer_mean'], yerr=four_pass_pos['hmscer_err'], fmt='o', label = r"4-pass data, $\pi^+$", color='r', elinewidth=1.5, capsize=0)
plt.errorbar(five_pass_neg['z'], five_pass_neg['hmscer_mean'], yerr=five_pass_neg['hmscer_err'], fmt='o', label = r"5-pass data, $\pi^-$", color='b', elinewidth=1.5, capsize=0)
plt.errorbar(five_pass_pos['z'], five_pass_pos['hmscer_mean'], yerr=five_pass_pos['hmscer_err'], fmt='o', label = r"5-pass data, $\pi^+$", color='g', elinewidth=1.5, capsize=0)
plt.xlabel(r"$z$", fontsize=14)
plt.ylabel("Fitted mean", fontsize=14)
plt.title(r"H.cer.npeSum fitted mean vs $z$", fontsize=16)
plt.legend(fontsize=8)
plt.grid(True, alpha=0.5)
plt.tight_layout()
```
