```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 


z36_df = pd.read_csv("../4pass//z0.36/th2.0/pi-/LH2_ratios.csv")
z50_df = pd.read_csv("../4pass//z0.5/th2.0/pi-/LH2_ratios.csv")
z67_df = pd.read_csv("../4pass//z0.67/th2.0/pi-/LH2_ratios.csv")

phi_z36 = z36_df[z36_df["variable"]=="$\phi_{pq}$"]
phi_z50 = z50_df[z50_df["variable"]=="$\phi_{pq}$"]
phi_z67 = z67_df[z67_df["variable"]=="$\phi_{pq}$"]

xsec_z36 = phi_z36["ratio"] * 1.9004677020359796E-003
xsec_z36_err = phi_z36["ratio_err"] * 1.9004677020359796E-003

xsec_z50 = phi_z50["ratio"] * 1.4056776245709084E-003
xsec_z50_err = phi_z50["ratio_err"] * 1.4056776245709084E-003

xsec_z67 = phi_z67["ratio"] * 9.2047611149291249E-004
xsec_z67_err = phi_z67["ratio_err"] * 9.2047611149291249E-004

fig, ax = plt.subplots(figsize=(17, 7))

plt.errorbar(
    phi_z36["bin_center"],
    xsec_z36,
    yerr=xsec_z36_err,
    fmt='o',
    label=r'$z = 0.36$'
)

plt.errorbar(
    phi_z50["bin_center"],
    xsec_z50,
    yerr=xsec_z50_err,
    fmt='o',
    label=r'$z = 0.5$'
)

plt.errorbar(
    phi_z67["bin_center"],
    xsec_z67,
    yerr=xsec_z67_err,
    fmt='o',
    label=r'$z = 0.67$'
)      

plt.xlabel(r"$\phi_{pq}$ (rad)", fontsize=14)
plt.ylabel(r"$\frac{d\sigma(x,Q^2,z,p_T)}{dE_e d\Omega_e dP_{\pi} d\Omega_{\pi}}$ ($\mu$b/GeV/sr$^2$)", fontsize=14)
plt.title("LH2 Semi-Inclusive $\pi^-$ cross section vs. $\\phi_{pq}$ (4pass, $Q^2 = 3.3, x_{bj} = 0.25, \\theta_{pq} = 2.0 $)", fontsize=16)

plt.grid()
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
```
