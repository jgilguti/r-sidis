```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import matplotlib.lines as mlines
from scipy.optimize import curve_fit

th0p8z50_df = pd.read_csv("../5pass/th-0.8/z0.5/pi-/LH2_ratios.csv")
th0p8z67_df = pd.read_csv("../5pass/th-0.8/z0.67/pi-/LH2_ratios.csv")
th2z36_df   = pd.read_csv("../5pass/th2.0/z0.36/pi-/LH2_ratios.csv")
th2z50_df   = pd.read_csv("../5pass/th2.0/z0.5/pi-/LH2_ratios.csv")
th2z67_df   = pd.read_csv("../5pass/th2.0/z0.67/pi-/LH2_ratios.csv")

def select_phi(df):
    return df[(df["variable"]=="$\phi_{pq}$") & (df["ratio_err"] < 0.09)]

phi_th0p8z50 = select_phi(th0p8z50_df)
phi_th0p8z67 = select_phi(th0p8z67_df)
phi_th2z36 = select_phi(th2z36_df)
phi_th2z50 = select_phi(th2z50_df)
phi_th2z67 = select_phi(th2z67_df)


xsec_th0p8z50     = phi_th0p8z50["ratio"]     * 4.8671936668655876E-003/0.18817829762735644 * 20
xsec_th0p8z50_err = phi_th0p8z50["ratio_err"] * 4.8671936668655876E-003/0.18817829762735644 * 20

xsec_th0p8z67     = phi_th0p8z67["ratio"]     * 3.2991551348072069E-003/7.0987298299679549E-002 * 20
xsec_th0p8z67_err = phi_th0p8z67["ratio_err"] * 3.2991551348072069E-003/7.0987298299679549E-002 * 20

xsec_th2z36     = phi_th2z36["ratio"]     * 6.2286428150942199E-003/0.46545583400647678 * 20
xsec_th2z36_err = phi_th2z36["ratio_err"] * 6.2286428150942199E-003/0.46545583400647678 * 20

xsec_th2z50     = phi_th2z50["ratio"]     * 4.6070048058393879E-003/0.17820992157316212 * 20
xsec_th2z50_err = phi_th2z50["ratio_err"] * 4.6070048058393879E-003/0.17820992157316212 * 20

xsec_th2z67     = phi_th2z67["ratio"]     * 3.0167926096160780E-003/6.4944992539682556E-002 * 20
xsec_th2z67_err = phi_th2z67["ratio_err"] * 3.0167926096160780E-003/6.4944992539682556E-002 * 20


datasets = {
    "z = 0.50 (th=-0.8)": (phi_th0p8z50, xsec_th0p8z50, xsec_th0p8z50_err),
    "z = 0.67 (th=-0.8)": (phi_th0p8z67, xsec_th0p8z67, xsec_th0p8z67_err),
    "z = 0.36 (th=2.0)": (phi_th2z36, xsec_th2z36, xsec_th2z36_err),
    "z = 0.50 (th=2.0)": (phi_th2z50, xsec_th2z50, xsec_th2z50_err),
    "z = 0.67 (th=2.0)": (phi_th2z67, xsec_th2z67, xsec_th2z67_err),
}

pt_bins = [0.05, 0.15, 0.25, 0.35]


fig, axes = plt.subplots(2, 2, figsize=(22, 8), sharey=True)
# axes = axes.flatten()

for ax, ptbc in zip(axes.flatten(), pt_bins):

    for label, (df_phi, xsec, xsec_err) in datasets.items():

        mask = df_phi["pt_bin_center"] == ptbc
        df_pt = df_phi[mask]

        ax.errorbar(
            df_pt["bin_center"],
            xsec[mask],
            yerr=xsec_err[mask],
            fmt='o',
            label=label
        )

    ax.set_title(fr"$p_T$ = {ptbc} GeV", fontsize=14)
    ax.set_xlabel(r"$\phi_{pq}$ (rad)", fontsize=12)
    ax.grid()
    ax.legend(fontsize=6)

ax.set_ylabel(
    # r"$\frac{d\sigma(x,Q^2,z,p_T)}{dE_e d\Omega_e dP_{\pi} d\Omega_{\pi}}$ ($\mu$b/GeV/sr$^2$)",
    r"$M(z,p_T,\phi_{pq})$",
    fontsize=13
)

plt.suptitle(
    r"LH2 Semi-Inclusive $\pi^-$ multiplicities vs. $\phi_{pq}$" "\n"
    r"(5pass, $Q^2 = 3.3,\ x_{bj}=0.25,\ \theta_{pq}=-0.8, 2.0^\circ)$",
    fontsize=17
)

plt.tight_layout(rect=[0, 0, 1, 1])
plt.show()
```

```python
# plt.style.use(['science','grid','no-latex'])
plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams['xtick.labelsize'] = 16
plt.rcParams['ytick.labelsize'] = 16

def phi_fit(phi, M0, A, B):
    return M0 * (1 + A*np.cos(phi) + B*np.cos(2*phi))

fig, axes = plt.subplots(2, 2, figsize=(20, 10), sharey=True, constrained_layout=True)

marker_styles = {
    "-0.8": "o",
    "2.0": "s",
}

for ax, ptbc in zip(axes.flatten(), pt_bins):

    for zval in ["0.36", "0.50", "0.67"]:
        phi_all, y_all, yerr_all = [], [], []

        for key, (df_phi, xsec, xsec_err) in datasets.items():

            # Parse the key: "z = 0.50 (th=-0.8)"
            z = key.split()[2]               # → "0.50"
            th = key.split("(th=")[1][:-1]   # → "-0.8"


            if z != zval:
                continue

            mask = df_phi["pt_bin_center"] == ptbc
            df_pt = df_phi[mask]

            if len(df_pt) == 0:
                continue

            phi_vals = df_pt["bin_center"].to_numpy()
            y_vals   = xsec[mask].to_numpy()
            y_errs   = xsec_err[mask].to_numpy()

            # Append to combined arrays
            phi_all.extend(phi_vals)
            y_all.extend(y_vals)
            yerr_all.extend(y_errs)

            # Plot individual θ datasets (same color, diff markers)
            ax.errorbar(
                phi_vals,
                y_vals,
                yerr=y_errs,
                fmt=marker_styles[th],
                markersize=4,
                color=plt.get_cmap('Dark2')(["0.36", "0.50", "0.67"].index(zval)),
                alpha=0.5,
                label=fr"z={zval}, $\theta_{{pq}}={th}^\circ$"
            )

        if len(phi_all) < 3:
            continue

        # Convert lists to arrays
        phi_all = np.array(phi_all)
        y_all = np.array(y_all)
        yerr_all = np.array(yerr_all)

        # ------------------------------
        # Fit combined θ data for this z
        # ------------------------------
        p0 = [np.mean(y_all), 1, 0.05]
        try:
            popt, pcov = curve_fit(phi_fit, phi_all, y_all, sigma=yerr_all, p0=p0)
            M0_fit, A_fit, B_fit = popt
            perr = np.sqrt(np.diag(pcov))
            M0_err, A_err, B_err = perr       
        except RuntimeError:
            M0_fit, A_fit, B_fit = np.nan, np.nan, np.nan
            M0_err = A_err = B_err = np.nan
        
        M0, A, B = popt

        # fit_text = (
        #     rf"${M0:.3f}[1 + {A:.3f} \cos\phi + {B:.3f}\cos(2\phi)]$" "\n"
        # ) 

        # ax.text(
        #     0.04, 0.97,
        #     fit_text,
        #     transform=ax.transAxes,
        #     fontsize=11,
        #     va="top",
        #     ha="left",
        #     family="monospace",
        #     bbox=dict(boxstyle="round", facecolor="white", alpha=0.25, edgecolor="none")
        # )

        print(f"\n Fit results for z = {zval}, pT = {ptbc} GeV:")
        print(f"M0 = {M0_fit:.5f} ± {M0_err:.5f}")
        print(f"A  = {A_fit:.5f} ± {A_err:.5f}")
        print(f"B  = {B_fit:.5f} ± {B_err:.5f}")
        print("===============================================")




        # Plot fit curve once
        phi_smooth = np.linspace(min(phi_all), max(phi_all), 300)
        ax.plot(
            phi_smooth,
            phi_fit(phi_smooth, M0_fit, A_fit, B_fit),
            linestyle="-",
            color=plt.get_cmap('Dark2')(["0.36", "0.50", "0.67"].index(zval)),
            linewidth=2.2,
            label=fr"Fit (z={zval}): $M_0={M0_fit:.3f}$, $A={A_fit:.3f}$, $B={B_fit:.3f}$"
        )

    legend_handles = [
        mlines.Line2D(
            [], [], color=plt.get_cmap('Dark2')(["0.36", "0.50", "0.67"].index(z)), marker='o', linestyle='',
            label=f"z = {z}"
        )
        for z in ["0.36", "0.50", "0.67"]
    ]

    # Place the legend centered below all subplots
    fig.legend(
        handles=legend_handles,
        loc="lower center",
        ncol=3,
        fontsize=14,
        frameon=False
    )

    # fig.subplots_adjust(bottom=0.52)

    ax.set_title(fr"$p_T = {ptbc}$ GeV", fontsize=18)
    ax.set_ylabel(r"$M(z,p_T,\phi_{pq})$",fontsize=15)
    ax.set_xlabel(r"$\phi_{pq}$ (rad)", fontsize=15)
    # ax.legend(fontsize=8)
    ax.grid(alpha=0.4)

ax.set_ylabel(r"$M(z,p_T,\phi_{pq})$")

for i, ax in enumerate(axes.flat):
    # compute fit parameters here: M0, A, B, etc.

    fit_text = (
        rf"$M_0[1 + A\cos\phi + B\cos(2\phi)]$" "\n"
        rf"$M_0={M0:.3f},\ A={A:.3f},\ B={B:.3f}$"
    )




# plt.style.available
```
