```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 

plt.rcParams['xtick.labelsize'] = 16
plt.rcParams['ytick.labelsize'] = 16

df = pd.read_csv('Wpeak_vs_Pdp.csv')

fig, ax = plt.subplots(figsize=(16,7))

plt.errorbar(df['Pdp_bin'], df['Wpeak_mean'], yerr=df['Wpeak_stddev'], fmt='o', ecolor='lightgray', elinewidth=3, capsize=0)
plt.axhline(y=0.938, color='r', linestyle='--', label='Proton mass (0.938 GeV/c)')

plt.title(r'Mean $W_{peak}$ vs P.gtr.dp', fontsize=20)
plt.xlabel(r'P.gtr.dp (%)', fontsize=18)
plt.ylabel(r'Mean $W_{peak}$ (GeV/c)', fontsize=18)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```
